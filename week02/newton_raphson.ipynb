# Newton-Raphson method: the TLDR

Given a function $f(x)$, Newton-Raphson method (or simply Newton's method) finds any $x_r$ such that $f(x_r)=0$.

The method uses the approximation

$$ 
x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}
$$
where $f'$ is the first derivative of $f(x)$ and the initial value $x_0$ is a *reasonable* guess for $x_r$. 

The method converges as

$$\lim_{n\rightarrow\infty} x_{n+1} = x_r$$

The method does not work with any function $f$ but it is usually realiable with polynomials.

## Special application: square root

The objective is to find the square root of a given number $\sqrt{a}$. We define a function
$$
f(x) = x^2-a
$$
and apply the Newton-Raphson method:

\begin{align}
x_{n+1} & = x_n - \frac{x^2-a}{2x} \\
        & = \frac{1}{2} \left ( x_n + \frac{a}{x_n} \right )
\end{align}

We expect

$$
\lim_{n\rightarrow\infty} {|x_{n+1}^2-a|} = 0
$$

and in practice we accept $x_{n_+1} = \sqrt{a}$ when

$$ |x_{n+1}^2 -a | \leq\epsilon $$

where $\epsilon$ is a sufficiently small value but not zero.
