# Newton-Raphson method: the TLDR

Given a function $f(x)$, Newton-Raphson method (or simply Newton's method) finds any $x_r$ such that $f(x_r)=0$.

The method uses the approximation

$$
x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}
$$

where $f'$ is the first derivative of $f(x)$ and the initial value $x_0$ is a _reasonable_ guess for $x_r$.

The method converges as

$$\lim_{n\rightarrow\infty} x_{n+1} = x_r$$

The method does not work with any function $f$ but it is usually realiable with polynomials.

## Special application: square root

The objective is to find the square root of a given number $\sqrt{a}$. We define a function

$$
f(x) = x^2-a
$$

and apply the Newton-Raphson method:

\begin{align}
x\_{n+1} & = x_n - \frac{x^2-a}{2x} \\
& = \frac{1}{2} \left ( x_n + \frac{a}{x_n} \right )
\end{align}

We expect

$$
\lim_{n\rightarrow\infty} {|x_{n+1}^2-a|} = 0
$$

and in practice we accept $x_{n_+1} = \sqrt{a}$ when

$$ |x\_{n+1}^2 -a | \leq\epsilon $$

where $\epsilon$ is a sufficiently small value but not zero.


```python
def newton_sqrt(a):
    """Calculate the square root of a number using Newton's method."""
    # Guard statements to handle edge cases
    if a < 0:
        raise ValueError("Cannot compute square root of a negative number.")
    elif a == 0:
        return 0

    # Initial guess and parameters for convergence
    x = a / 2.0  
    # Define the tolerance level for convergence
    tolerance = 0.0001 
    # Define a maximum number of iterations to prevent infinite loops 
    stop_after = 1_000  
    # Initialize the iteration count
    iterations_count = 0

    while abs(x * x - a) > tolerance and iterations_count < stop_after:
        # Update guess using Newton's formula:
        x = (x + a / x) / 2.0
        iterations_count += 1
    # Done
    return x
```
