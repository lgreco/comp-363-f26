# String alignment

## Notation

Strings $X_m$ and $Y_n$ of length $m$ and $n$ respectively, with $m\geq n > 0$. 

Each string comprises symbols $x_i$ and $y_j$ with $0\leq i <m$ and $0\leq j < n$:

\begin{align}
X_m & = x_0\, x_1\, \ldots\, x_{m-2}\, x_{m-1} \\
Y_n & = y_0\, y_1\, \ldots\, y_{n-2}\, y_{n-1}
\end{align}

For example,

\begin{align}
X_7 & = \texttt{BICYCLE}, && & x_0 & = \texttt{B},&\, x_1 = \texttt{I},\,\ldots \\
Y_5 & = \texttt{CYCLE},  &&  & y_0 & = \texttt{C},&\, y_1 = \texttt{Y},\,\ldots
\end{align}

$X_i$, where $0\leq i < m$, is a substring with the first $i$ symbols of $X_m$: $x_0\, x_1\,\ldots\, x_{i-1}$. Conversely, $Y_j$, where $0\leq j <n$ is a substring with the first $j$ symbols of $Y_n$: $y_0\,\ldots\, y_j$. Using the previous example, 

\begin{align}
X_3 &= \texttt{BIC} && \\
Y_3 &= \texttt{CYC} &&
\end{align}

An alignment of $(X_m, Y_n)$ is the pair of strings of equal length, $(\overline{X}_L, \overline{Y}_L)$, with $\max(m,n) \leq L\leq m+n$.

For example,

\begin{align}
(\overline X_7, \overline Y_7) = ( &\texttt{BICYCLE}, \\ &\texttt{CYCLE--})
\end{align}
is an alignment of the earlier $X_7$ and $Y_5$ strings. In the strings above, "$\texttt{-}$" denotes a space.

Another possible alignment is

\begin{align}
( &\texttt{BICYCLE}, \\ &\texttt{--CYCLE})
\end{align}

This particular alignment is *optimal* because it maximizes the number of identical overlapping symbols between the two strings.

## The problem

Given two strings $X_m$ and $Y_n$ find their optimal alignment $(\overline X_L, \overline Y_L)$.

## The challenge

First, we need to define how to measure optimality. A simple rule is to maximine the number of similar letters lining up as we see in
\begin{align}
( &\texttt{BICYCLE}, \\ &\texttt{--CYCLE})
\end{align}

Second, we have to find every possible alignment between the two strings and find the one with the best optimality measure. And this is a **big** problem. There are

$$
\sum_{L = \max(m,n)}^{m+n} = \frac{L!}{(m+n−L)!(L−m)!(L−n)!}
$$

For strings ``BICYCLE`` and ``CYCLE`` that's 18,348 combinations. If our computer can evaluate $10^6$ combinations per second, we can find the optimal one in under 1/50th of a second.

If however, $m=100$ and $n=50$, the total number of possible combinations is $4.941\times 10^{32}$. The same computer evaluating $10^6$ combinations per second, would take $1.57\times 10^{19}$ years to find the optimal combination. For reference, the universe is about $1.38\times 10^{10}$ years old. Things get even worst for larger strings. [Sequence alignment](https://en.wikipedia.org/wiki/Sequence_alignment) in biology uses large strings and we can't wait for ever to get a match.

**Obviously there is a solution to both challenges above!**

## Solution to first challenge

In an alignment $(\overline X_L, \overline Y_L)$ of strings $(X_m, Y_n)$ we define the following two measuments.

### Match bonus / mismatch malus

$$
a_{\overline x_k\,\overline y_k} = \begin{cases} 0\ & \text{when}\ \overline x_k = \overline y_k \\ p \geq 1 \ & \text{when}\ \overline x_k \neq \overline y_k \end{cases}
$$

### Gap measure

$$
a_{\text{gap}} = g\geq 1\ \text{when either}\ \overline x_k = \texttt{'-'}\ \text{or}\ \overline y_k = \texttt{'-'}
$$

# Measuring optimality 

In the measurements above, $\overline x_k$ and $\overline y_k$ are the $k$-th symbols in the aligned strings $\overline X_L$ and $\overline Y_L$ respectively; $p$ and $g$ are positive integers. Using the earlier example, and assuming $g=1$ and $p=3$.

\begin{align}
k &&\rightarrow\quad               & 0 && 1 && 2 && 3 && 4 && 5 && 6 \\
\overline X_7 &&\rightarrow\quad   & \texttt{B} && \texttt{I} && \texttt{C} && \texttt{Y} && \texttt{C} && \texttt{L} && \texttt{E} \\
\overline Y_7 &&\rightarrow\quad   & \texttt{-} && \texttt{-} && \texttt{C} && \texttt{Y} && \texttt{C} && \texttt{L} && \texttt{E}\\
\text{penalty} &&\rightarrow\quad  & 1 && 1 && 0 && 0 && 0 && 0 && 0
\end{align}
The total penalty for the alignemnt above is $2$.

For the same values of $p$ and $g$ but for strings `DOGS` and `CATS` we may be tempted to consider the following alignment that allows the overlap of at least one character,

\begin{align}
k \rightarrow\quad              & 0 && 1 && 2 && 3\\
\overline X_4 \rightarrow\quad  & \texttt D && \texttt O && \texttt G && \texttt S\\
\overline Y_4 \rightarrow\quad  & \texttt C && \texttt A && \texttt T && \texttt S\\
\text{penalty} \rightarrow\quad & 3 &&  3 &&  3 &&  0
\end{align}

with a total penalty of $9$. However, there is a better alignment with a lower total penalty of 8.


\begin{align}
k \rightarrow\quad              & 0 && 1 && 2 && 3 && 4&& 5 && 6 && 7\\
\overline X_8 \rightarrow\quad  & \texttt D && \texttt O && \texttt G && \texttt S && \texttt - && \texttt - && \texttt - && \texttt -\\
\overline Y_8 \rightarrow\quad  & \texttt - && \texttt - && \texttt - && \texttt - && \texttt C && \texttt A && \texttt T && \texttt S\\
\text{penalty} \rightarrow\quad & 1 &&  1 &&  1 &&  1  &&  1 &&  1 &&  1 &&  1
\end{align}


## Solution to the second challenge

The solution to the second challenge is actually **easy.** Because we assume that somehow it appears before. So for strings $X_7 =\texttt{BICYCLE}$ and $Y_5=\texttt{CYCLE}$, somehow the optimal solution is given to us (miraculously, if you wish). After we recover from the shock of such unexpected gift, we focus on the last column of the alignment $(\overline X_7, \overline Y_7)$ below.


\begin{align}
k &&\rightarrow\quad               & 0 && 1 && 2 && 3 && 4 && 5 && 6 \\
\overline X_7 &&\rightarrow\quad   & \texttt{B} && \texttt{I} && \texttt{C} && \texttt{Y} && \texttt{C} && \texttt{L} && \texttt{\color{maroon}E} \\
\overline Y_7 &&\rightarrow\quad   & \texttt{-} && \texttt{-} && \texttt{C} && \texttt{Y} && \texttt{C} && \texttt{L} && \texttt{\color{maroon}E}\\
\text{penalty} &&\rightarrow\quad  & 1 && 1 && 0 && 0 && 0 && 0 && 0
\end{align}

That last column can be in only one of three possible forms:

$$\left |\begin{matrix} x_6 \\ y_4 \end{matrix}\right |,\quad
\left |\begin{matrix} x_6 \\ \texttt - \end{matrix}\right |,\,\text{or}\quad
\left |\begin{matrix}\texttt - \\ y_4 \end{matrix}\right |
$$

In the example here, it's in the form $\left |\begin{matrix} x_6 \\ y_4 \end{matrix}\right |$ and specifically $\left |\begin{matrix}\texttt E \\ \texttt E \end{matrix}\right |$. We can argue -- and prove, later -- that the remaining alignment 


\begin{align}
k &&\rightarrow\quad               & 0 && 1 && 2 && 3 && 4 && 5 \\
\overline X_6 &&\rightarrow\quad   & \texttt{B} && \texttt{I} && \texttt{C} && \texttt{Y} && \texttt{C} && \texttt{L}  \\
\overline Y_6 &&\rightarrow\quad   & \texttt{-} && \texttt{-} && \texttt{C} && \texttt{Y} && \texttt{C} && \texttt{L} \\
\text{penalty} &&\rightarrow\quad  & 1 && 1 && 0 && 0 && 0 && 0 
\end{align}

is the optimal alignment of $X_6=\texttt{BICYCL}$ and $Y_4=\texttt{CYCL}$.

In general, given an optimal alignment $(\overline X_L, \overline Y_L)$ of two strings $(X_m, Y_n)$, the alignment $(\overline X_{L-1}, \overline Y_{L-1})$ is also the optimal alignment of either $(X_{m-1}, Y_{n-1})$ or $(X_{m-1}, Y_n)$ or $(X_m, Y_{n-1})$. It can be proven, later, that 

$$
(\overline X_{L-1}, \overline Y_{L-1})\ \text{is optimal alignment of}\ 
\begin{cases}
(X_{m-1}, Y_{n-1})\ \text{if}\ \left |\begin{matrix}\overline x_{L-1} \\ \overline y_{L-1} \end{matrix}\right | = \left |\begin{matrix} x_{m-1} \\ y_{n-1} \end{matrix}\right | \\ \\
(X_{m-1}, Y_{n})\ \text{if}\ \left |\begin{matrix}\overline x_{L-1} \\ \overline y_{L-1} \end{matrix}\right | = \left |\begin{matrix} x_{m-1} \\ \texttt - \end{matrix}\right |  \\ \\
(X_{m}, Y_{n-1})\ \text{if}\ \left |\begin{matrix}\overline x_{L-1} \\ \overline y_{L-1} \end{matrix}\right | = \left |\begin{matrix} \texttt - \\ y_{n-1} \end{matrix}\right | 
\end{cases}
$$

This is a notational nightmare, so here's a simple rule to decipher things. An uppercase variable, $X$ for example, is a string. If the string has length $m$, we write $X_m$. A lowercase variable, $x$ is a symbol (a single character) within string $X$. The first character in $X_m$ is $x_0$ and the last character is $x_{m-1}$.
