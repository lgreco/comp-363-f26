## Notation

**Heads up:** it's notation that makes this process challenging. The ideas behind it are quite simple. So let's clarify the notation used here upfront.

**An uppercase variable, $X$ for example, is a string. If the string has length $m$, we write $X_m$. A lowercase variable, $x$ is a symbol (a single character) within string $X$. The first character in $X_m$ is $x_0$ and the last character is $x_{m-1}$.**

Notation helps abbreviate things. For example, instead of writing *"a string with 7 characters"*, we denote $X_7$. Instead of saying *"the 3rd character of the string"*, we write $x_2$.

For our analysis we will consider two strings $X_m$ and $Y_n$, with $m\geq n > 0$. (In more words we would have said *"string $X$ with length $m$ and string $Y$ with length $n$")*.

Each string comprises symbols $x_i$ and $y_j$ with $0\leq i <m$ and $0\leq j < n$:

\begin{align}
X_m & = x_0\, x_1\, \ldots\, x_{m-2}\, x_{m-1} \\
Y_n & = y_0\, y_1\, \ldots\, y_{n-2}\, y_{n-1}
\end{align}

For example,

\begin{align}
X_7 & = \texttt{BICYCLE}, && & x_0 & = \texttt{B},&\, x_1 = \texttt{I},\,\ldots \\
Y_5 & = \texttt{CYCLE},  &&  & y_0 & = \texttt{C},&\, y_1 = \texttt{Y},\,\ldots
\end{align}

$X'_i$, where $0\leq i \leq m$, is a prefix with the first $i$ symbols of $X_m$: $x_0\, x_1\,\ldots\, x_{i-1}$. 

Conversely, $Y'_j$, where $0\leq \leq <n$ is a prefix with the first $j$ symbols of $Y_n$: $y_0\,\ldots\, y_j$. Simply, a prefix of a string is a substring that begins at the first character of the string. In these notes, we use the *prime* ($'$) to indicate the prefix of a string, e.g., $X'$ is a prefix of $X$. Using the previous example with $X_m=\texttt{BICYCLE}$,

\begin{align}
X'_0 & = \texttt{""},\,\ \text{(the empty string)} \\
X'_1 & = \texttt{B} \\
X'_2 & = \texttt{BI} \\
X'_3 & = \texttt{BIC} \\
X'_4 & = \texttt{BICY} \\
X'_5 & = \texttt{BICYC} \\
X'_6 & = \texttt{BICYCL} \\
X'_7 & = \texttt{BICYCLE} \\
\end{align}

(And yes, to make things a bit more confusing: $X'_7=X_7$, a string is a prefix of itself!)

Next, we define the alignment of $(X_m, Y_n)$. Their alignment is a pair of strings of equal length $L$ such that 

$$
\max(m,n) \leq L\leq m+n
$$ 
For example,

$$
\left |{\texttt B\atop\texttt C} \right |{\texttt I\atop\texttt Y } \left | {\texttt C\atop\texttt - } \right |{\texttt Y\atop\texttt C }  \left | {\texttt C\atop\texttt L } \right |{\texttt L\atop\texttt E }  \left | {\texttt E\atop\texttt - } \right |
$$
is an alignment of the earlier $X_7$ and $Y_5$ strings. In the strings above, "$\texttt{-}$" denotes a space.

Another possible alignment is

$$
\left |{\texttt B\atop\texttt -} \right |{\texttt I\atop\texttt - } \left | {\texttt C\atop\texttt C } \right |{\texttt Y\atop\texttt Y }  \left | {\texttt C\atop\texttt C } \right |{\texttt L\atop\texttt L }  \left | {\texttt E\atop\texttt E } \right |
$$

This particular alignment is *optimal* because it maximizes the number of identical overlapping symbols between the two strings.  

We denote the **optimal alignment** of two strings as $\overline{(X_m, Y_n)}_L$.


## The problem

Given two strings $X_m$ and $Y_n$ find their optimal alignment $\overline{(X_m, Y_n)}_L$.

## The challenge

First, we need to define how to measure optimality. A simple rule is to maximine the number of similar letters lining up as we see in

$$
\left |{\texttt B\atop\texttt -} \right |{\texttt I\atop\texttt - } \left | {\texttt C\atop\texttt C } \right |{\texttt Y\atop\texttt Y }  \left | {\texttt C\atop\texttt C } \right |{\texttt L\atop\texttt L }  \left | {\texttt E\atop\texttt E } \right |
$$

Second, we have to find every possible alignment between the two strings and then find the one with the best optimality measure. And this is a **big** problem. There are

$$
\sum_{L = \max(m,n)}^{m+n} = \frac{L!}{(m+n−L)!(L−m)!(L−n)!}
$$

possible alignments.

For strings ``BICYCLE`` and ``CYCLE`` that's 18,348 combinations. If our computer can evaluate $10^6$ combinations per second, we can find the optimal one in under 1/50th of a second.

If however, $m=100$ and $n=50$, the total number of possible combinations is $4.941\times 10^{32}$. The same computer evaluating $10^6$ combinations per second, would take $1.57\times 10^{19}$ years to find the optimal combination. For reference, the universe is about $1.38\times 10^{10}$ years old. Things get even worst for larger strings. [Sequence alignment](https://en.wikipedia.org/wiki/Sequence_alignment) in biology uses large strings and we can't wait for ever to get a match.

**Obviously there is a solution to both challenges above!**

## Solution to first challenge

In an alignment $\overline{(X_m, Y_n)}_L$  we define the following two measuments.

### Match bonus / mismatch malus

Given two symbols in an alignment column $\left | x\atop y \right |$, we reward (bonus) similarity and penalize (malus) a mismatch:
$$
a_{xy} = \begin{cases} 0\ & \text{when}\ x = y \\ p \geq 1 \ & \text{when}\  x \neq y \end{cases}
$$

### Gap penalty

We also charge a penalty for every alignment column with a space (gap), like $\left | x\atop \texttt{-} \right |$ or $\left | \texttt{-} \atop y \right |$:

$$
a_{\text{gap}} = 
  g\geq 1
$$

There are no alignment columns with only gaps in them: $\left | \texttt{-}\atop \texttt{-} \right |$.

# Measuring optimality 

Using the earlier example, and assuming $g=1$ for gap penalty and $p=3$ for mismatch penalty, we get:

\begin{align}
k &&\rightarrow\quad               & 0 && 1 && 2 && 3 && 4 && 5 && 6 && (k\text{ is the index of the alignment columns)}\\
\overline X_7 &&\rightarrow\quad   & \texttt{B} && \texttt{I} && \texttt{C} && \texttt{Y} && \texttt{C} && \texttt{L} && \texttt{E}  && (\overline X_7\text{ is the aligned version of the first input string)}\\
\overline Y_7 &&\rightarrow\quad   & \texttt{-} && \texttt{-} && \texttt{C} && \texttt{Y} && \texttt{C} && \texttt{L} && \texttt{E}  && (\overline Y_7\text{ is the aligned version of the second input string)} \\
\text{penalty} &&\rightarrow\quad  & 1 && 1 && 0 && 0 && 0 && 0 && 0
\end{align}

The total penalty for the alignemnt above is $2$. 

For the same values of $p$ and $g$ but for strings `DOGS` and `CATS` we may be tempted to consider the following alignment that allows the overlap of at least one character,

\begin{align}
k \rightarrow\quad              & 0 && 1 && 2 && 3\\
\overline X_4 \rightarrow\quad  & \texttt D && \texttt O && \texttt G && \texttt S\\
\overline Y_4 \rightarrow\quad  & \texttt C && \texttt A && \texttt T && \texttt S\\
\text{penalty} \rightarrow\quad & 3 &&  3 &&  3 &&  0
\end{align}

with a total penalty of $9$. However, there is a better alignment with a lower total penalty of 8.


\begin{align}
k \rightarrow\quad              & 0 && 1 && 2 && 3 && 4&& 5 && 6 && 7\\
\overline X_8 \rightarrow\quad  & \texttt D && \texttt O && \texttt G && \texttt S && \texttt - && \texttt - && \texttt - && \texttt -\\
\overline Y_8 \rightarrow\quad  & \texttt - && \texttt - && \texttt - && \texttt - && \texttt C && \texttt A && \texttt T && \texttt S\\
\text{penalty} \rightarrow\quad & 1 &&  1 &&  1 &&  1  &&  1 &&  1 &&  1 &&  1
\end{align}


## Solution to the second challenge

The solution to the second challenge is actually **easy.** It appears, magically, before our eyes. So for strings $X_7 =\texttt{BICYCLE}$ and $Y_5=\texttt{CYCLE}$, somehow the optimal solution is given to us (miraculously, if you wish). After we recover from the shock of such unexpected gift, we focus on the last column of the alignment $\overline{( X_7, Y_5)}$ below.


\begin{align}
k &&\rightarrow\quad               & 0 && 1 && 2 && 3 && 4 && 5 && 6 \\
\overline X_7 &&\rightarrow\quad   & \texttt{B} && \texttt{I} && \texttt{C} && \texttt{Y} && \texttt{C} && \texttt{L} && \texttt{\color{maroon}E} \\
\overline Y_7 &&\rightarrow\quad   & \texttt{-} && \texttt{-} && \texttt{C} && \texttt{Y} && \texttt{C} && \texttt{L} && \texttt{\color{maroon}E}\\
\text{penalty} &&\rightarrow\quad  & 1 && 1 && 0 && 0 && 0 && 0 && 0
\end{align}

That last column can be in only one of three possible forms:

$$\left |\begin{matrix} x_6 \\ y_4 \end{matrix}\right |,\quad
\left |\begin{matrix} x_6 \\ \texttt - \end{matrix}\right |,\,\text{or}\quad
\left |\begin{matrix}\texttt - \\ y_4 \end{matrix}\right |
$$

In the example here, it's in the form $\left |\begin{matrix} x_6 \\ y_4 \end{matrix}\right |$ and specifically $\left |\begin{matrix}\texttt E \\ \texttt E \end{matrix}\right |$. We can argue -- and prove, later -- that the remaining alignment 


\begin{align}
k &&\rightarrow\quad               & 0 && 1 && 2 && 3 && 4 && 5 \\
\overline {X}_6 &&\rightarrow\quad   & \texttt{B} && \texttt{I} && \texttt{C} && \texttt{Y} && \texttt{C} && \texttt{L}  \\
\overline {Y}_6 &&\rightarrow\quad   & \texttt{-} && \texttt{-} && \texttt{C} && \texttt{Y} && \texttt{C} && \texttt{L} \\
\text{penalty} &&\rightarrow\quad  & 1 && 1 && 0 && 0 && 0 && 0 
\end{align}

is the optimal alignment of $X'_6=\texttt{BICYCL}$ and $Y'_4=\texttt{CYCL}$.

In general, the optimal alignment $\overline{( X_m, Y_n)}_L$ of two strings also contains the optimal alignment of either $(X'_{m-1}, Y'_{n-1})$ or $(X'_{m-1}, Y_n)$ or $(X_m, Y'_{n-1})$. It can be proven, later, that 

$$
\overline{(X_m, Y_n)}_L \text{ minus its last column}\left |\overline x_{L-1} \atop\overline y_{L-1} \right |\ \text{is optimal alignment of}\ 
\begin{cases}
(X'_{m-1}, Y'_{n-1})\, &\text{if}\ \left |\overline x_{L-1} \atop\overline y_{L-1} \right | = \left | x_{m-1}   \atop y_{n-1}   \right | \\ \\
(X'_{m-1}, Y_{n})\,    &\text{if}\ \left |\overline x_{L-1} \atop\overline y_{L-1} \right | = \left | x_{m-1}   \atop \texttt - \right |  \\ \\
(X_{m}, Y'_{n-1})\,    &\text{if}\ \left |\overline x_{L-1} \atop\overline y_{L-1} \right | = \left | \texttt - \atop y_{n-1}   \right | 
\end{cases}
$$

As mentioned earlier, notation is the most challenging part of this problem analysis. Let's refresh the notation used one more time. An uppercase variable, $X$ for example, is a string. If the string has length $m$, we write $X_m$. A lowercase variable, $x$ is a symbol (a single character) within string $X$. The first character in $X_m$ is $x_0$ and the last character is $x_{m-1}$. A prefix of $X$ is denoted with a prime: $X'$.

## Bicycles and cycles

We argued that if $\left|\texttt{BICYCLE}\atop\texttt{--CYCLE}\right|$
is an optimal alignment of `BICYCLE` and `CYCLE`, then
 $\left|\texttt{BICYCL}\atop\texttt{--CYCL}\right|$
is an optimal alignment of `BICYCL` and `CYCL`. 

On this token,   $\left|\texttt{BICYC}\atop\texttt{--CYC}\right|$
is an optimal alignment of `BICYC` and `CYC` and so on. In summary:

\begin{align*}

&& \left|\texttt{BICYCLE}\atop\texttt{--CYCLE}\right| && \text{is optimal alignment of} &&& X_7 = \texttt{BICYCLE}\ \text{and}\ Y_5=\texttt{CYCLE} &\\ \\
&& \left|\texttt{BICYCL}\atop\texttt{--CYCL}\right| && \text{is optimal alignment of} &&& X'_6 = \texttt{BICYCL}\ \text{and}\ Y'_4=\texttt{CYCL} &\\ \\
&& \left|\texttt{BICYC}\atop\texttt{--CYC}\right| && \text{is optimal alignment of} &&& X'_5 = \texttt{BICYC}\ \text{and}\ Y'_3=\texttt{CYC} & \\ \\
&& \left|\texttt{BICY}\atop\texttt{--CY}\right| && \text{is optimal alignment of} &&& X'_4 = \texttt{BICY}\ \text{and}\ Y'_2=\texttt{CY} & \\ \\
&& \left|\texttt{BIC}\atop\texttt{--C}\right| && \text{is optimal alignment of} &&& X'_3 = \texttt{BIC}\ \text{and}\ Y'_1=\texttt{C} & \\ \\
&& \left|\texttt{BI}\atop\texttt{--}\right| && \text{is optimal alignment of} &&& X'_2 = \texttt{BI}\ \text{and}\ Y'_0=\texttt{""} & \\ \\
&& \left|\texttt{B}\atop\texttt{-}\right| && \text{is optimal alignment of} &&& X'_1 = \texttt{B}\ \text{and}\ Y'_0=\texttt{""} & 

\end{align*}

## Where there is a pattern there is a process

An optimal alignment $(\overline X_L, \overline Y_L)$ can be rewritten as the concatenation of the immediately previous alignment plus the last column of the alignment:

\begin{align*}
\overline X_L & = \overline{X'}_{L-1}  + \overline x_{L-1} \\
\overline Y_L & = \overline{Y'}_{L-1}  + \overline y_{L-1}
\end{align*}

For example, 
$\left | \texttt{BICYLE}\atop \texttt{--CYLE} \right | = \left | \texttt{BICYL}\atop \texttt{--CYCL} \right | + \left | {\texttt E}\atop {\texttt E} \right |$. 

In general however, the last column could be any of the three distict possibilities: $\left | x_{m-1}\atop y_{n-1} \right |$ or $\left | x_{m-1}\atop \texttt - \right |$ or even $\left | \texttt -\atop y_{n-1} \right |$.

Each of these choices carries a cost and our objective is to keep to total cost down. So the cost of aligning $X_7$ and $Y_5$ can be selected from the following:
$$
\text{cost of aligning } X_7 \text{ and } Y_5 =\min\text{ of } \begin{cases}
    \text{cost for adding } {\left | x_{m-1}\atop y_{n-1}   \right |} \text{ to the optimal alignment of } X'_6 \text{ and } Y'_4 \\ \\
    \text{cost for adding } {\left | x_{m-1}\atop \texttt - \right |} \text{ to the optimal alignment of } X'_6 \text{ and } Y_5 \\ \\
    \text{cost for adding } {\left | \texttt -\atop y_{n-1} \right |} \text{ to the optimal alignment of } X_7  \text{ and } Y'_4 
    \end{cases}
$$

In a more compact notation where $P(7,5)$ is the cost of the optimal alignment for $X'_7$ and $Y'_5$, we can write

$$
P(7,5) = \min \left\{ P(6,4) + a_{x_6 y_4},\quad P(6,5) + a_\text{gap},\quad P(7,4) + a_\text{gap} \right \}
$$

where $P(6,5)$ is the cost of the optimal alignment between $X'_6$ and $Y'_5$ and $P(7,4)$ is the cost of the optimal alignment between $X'_7$ and $Y'_4$.

In general, we can write for the cost of aligning two strings of lengths $m$ and $n$ respectively.

$$ 
P(m,n) = \min \left \{ P(m-1,n-1) + a_{x_{m-1}\,y_{n-1}},\quad P(m-1,n)+a_\text{gap},\quad P(m,n-1)+a_\text{gap} \right \}
$$

By backtracking the computation, we get

\begin{align*}
P(i,j) & = \min \left \{ P(i-1,j-1) + a_{x_{i-1}\,y_{j-1}},\quad P(i-1,j)+a_\text{gap},\quad P(i,j-1)+a_\text{gap} \right \} \\
       & 0 \leq i < m \\
       & 0 \leq j < n
\end{align*}

And some point we need to deal with $i=0$ or $j=0$, so let's get these cases out of the way.

* $P(0,0)$ is the cost of optimally aligning two empty strings. There are no gaps needed and there are no penalties for mismatching characters, so this is trivial: $P(0,0)=0$.
* $P(i,0),\, i > 0$ is the cost of optimally aligning a string of length $i$ with an empty string. Such alignment requires only spaces at $a_\text{gap}$ per space, so $P(i,0)=i\times a_\text{gap}$.
* $P(0,j),\, j> 0$ is the cost of optimally aligning an empty string with a string of length $j$. Such alignment requires only spaces at $a_\text{gap}$ per space, so $P(0,j)=j\times a_\text{gap}$.

The last two cases above are similar and their only difference is which string is first. All three cases can be compacted to

$$ P(i,j) = (i+j)\times a_\text{gap},\ \text{ when } 0\leq (i+j)\leq 1 $$

These cases are extremely useful because they provide initial values for the computation

\begin{align*}
& \textbf{for } i=1\ldots m \\
& \quad\textbf{for } j=1\ldots n \\
& \quad\quad P(i,j) = \min \left \{ P(i-1,j-1) + a_{x_{i-1}\,y_{j-1}},\quad P(i-1,j)+a_\text{gap},\quad P(i,j-1)+a_\text{gap} \right \}
\end{align*}

At the end of these two loops we have the optimal alignment costs for all prefixe of the input strings $X_m$ and $Y_n$. The last term of the computation, $P(m,n)$ is the cost of the desired optimal alignment between the two full strings.

## Good news, bad news.

Good news is that we now know the cost of the optimal alignment of strings $X_m$ and $Y_n$ after only $(m+1)(n+1)$ computations. The bad news is that we don't know yet what that alignment looks like. Using the earlier example, we know that aligning `BICYCLE` and `CYCLE` has a cost of $2\times a_\text{gap}$. We don't know that the aligment actually is $\left | \texttt{BICYCLE}\atop\texttt{--CYCLE}\right |$.
