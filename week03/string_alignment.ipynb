# String alignment

## Notation

Strings $X_m$ and $Y_n$ of length $m$ and $n$ respectively, with $m\geq n > 0$. 

Each string comprises symbols $x_i$ and $y_j$ with $0\leq i <m$ and $0\leq j < n$:

\begin{align}
X_m & = x_0\, x_1\, \ldots\, x_{m-2}\, x_{m-1} \\
Y_n & = y_0\, y_1\, \ldots\, y_{n-2}\, y_{n-1}
\end{align}

For example,

\begin{align}
X_7 & = \texttt{BICYCLE}, && & x_0 & = \texttt{B},&\, x_1 = \texttt{I},\,\ldots \\
Y_5 & = \texttt{CYCLE},  &&  & y_0 & = \texttt{C},&\, y_1 = \texttt{Y},\,\ldots
\end{align}

$X_i$, where $0\leq i < m$, is a substring with the first $i$ symbols of $X_m$: $x_0\, x_1\,\ldots\, x_{i-1}$. Conversely, $Y_j$, where $0\leq j <n$ is a substring with the first $j$ symbols of $Y_n$: $y_0\,\ldots\, y_j$. Using the previous example, 

\begin{align}
X_3 &= \texttt{BIC} && \\
Y_3 &= \texttt{CYC} &&
\end{align}

An alignment of $(X_m, Y_n)$ is the pair of strings of equal length, $(\overline{X}_L, \overline{Y}_L)$, with $\min{(m,n)} \leq L\leq m+n$.

For example,

\begin{align}
(\overline X_7, \overline Y_7) = ( &\texttt{BICYCLE}, \\ &\texttt{CYCLE--})
\end{align}
is an alignment of the earlier $X_7$ and $Y_5$ strings. In the strings above, "$\texttt{-}$" denotes a space.

Another possible alignment is

\begin{align}
( &\texttt{BICYCLE}, \\ &\texttt{--CYCLE})
\end{align}

This particular alignment is *optimal* because it maximizes the number of identical overlapping symbols between the two strings.
